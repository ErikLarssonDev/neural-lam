```python
pr_stats_path='/mimer/NOBACKUP/groups/mlhighres/projects/detex/HCLIM_EC-Earth3-Veg/standardized_new/mean_std_remapped.pr_EUR-11_EC-Earth3-Veg_historical_r1i1p1f1_HCLIMcom-SMHI_HCLIM43-ALADIN_v1-r1_day_1951-2014_mm_day_noleap.npy' 
tas_stats_path='/mimer/NOBACKUP/groups/mlhighres/projects/detex/HCLIM_EC-Earth3-Veg/standardized_new/mean_std_remapped.tas_EUR-11_EC-Earth3-Veg_historical_r1i1p1f1_HCLIMcom-SMHI_HCLIM43-ALADIN_v1-r1_day_1951-2014_noleap.npy'

pr_mean, pr_std = np.load(pr_stats_path)
tas_mean, tas_std = np.load(tas_stats_path)
```

```python
for name, run_id in zip(names, run_ids):
    print(f"{name}: {run_id}")

    if name == "EDM_100":
        continue
    paths = []
    paths.append(glob.glob(f"wandb/*{run_id}*/files/test_ens_rmse.csv")[0])
    paths.append(glob.glob(f"wandb/*{run_id}*/files/test_spsk_ratio.csv")[0])
    paths.append(glob.glob(f"wandb/*{run_id}*/files/test_crps_ens.csv")[0])
    
    df = pd.concat(
        (pd.read_csv(p, header=None).assign(source=p.split('/')[-1]) for p in paths),
        ignore_index=True
    )
    
    df.columns = ["Precipitation", "Temperature", "source"]
    
    rows = [0, 2]
    
    df.loc[rows, "Precipitation"] = (
        df.loc[rows, "Precipitation"] * pr_std
    )
    
    df.loc[rows, "Temperature"] = (
        df.loc[rows, "Temperature"] * tas_std
    )
    
    sel = df.round(3).T
    print(sel)
    print()
    
    for row in ["Precipitation", "Temperature"]:
        values = sel.loc[row].astype(float)
        latex_row = " & ".join(f"{v:.3f}" for v in values) + r" \\"
        print(latex_row)
    print()
```
